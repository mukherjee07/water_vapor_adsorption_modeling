```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

#Reading the file
df = pd.read_csv('AA.csv',delimiter=',')
df2 = pd.read_csv('AR.csv',delimiter=',')
df3 = pd.read_csv('AAR.csv',delimiter=',')
df4 = pd.read_csv('Inner.csv',delimiter=',')
df5 = pd.read_csv('Outer.csv',delimiter=',')

#Reading the data
p = (df.iloc[:,0].values)
aa_h2o = df.iloc[:,1].values
aa_error = df.iloc[:,2].values
#2sigma
aa_error = 2*aa_error
uncertainty_aa = (aa_error) * 0.5

ar_h2o = df2.iloc[:,1].values
uncertainty_ar = df2.iloc[:,2].values

aar_h2o = df3.iloc[:,1].values
uncertainty_aar = df3.iloc[:,2].values

inner_h2o = df4.iloc[:,1].values
uncertainty_inner = df4.iloc[:,2].values

outer_h2o = df5.iloc[:,1].values
uncertainty_outer = df5.iloc[:,2].values

#CH4_Error = df.iloc[:,4].values
#print(Iast_co2)
#Doing the plotting
# set the font globally
#plt.rcParams.update({"font.family": "Times"})
marker_size = 55
fig1=plt.figure(dpi=800)
fig1.set_figwidth(15)
fig1.set_figheight(10)
plt.rc('xtick',labelsize=36)
plt.rc('ytick',labelsize=36)
#plt.rcParams.update({'errorbar.capsize':3})
#plotting for 100 K
#plt.plot(Pressure100,GP_100, marker = "H",color="r",Linestyle="solid",label="GP (T = 100 K)")

plt.plot(p,aa_h2o, marker = "o", markersize='27.5', 
         markeredgecolor='k',linewidth='10',color = "k", 
         markerfacecolor='orange',markeredgewidth='5',
         linestyle = " ",label="Alt Along",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, aa_h2o,  yerr = uncertainty_aa, 
             label="$\sigma_{CFC-GCMC}$",color = 'k', capsize=25,  # Length of the error bar caps
             elinewidth=10,  # Thickness of the error bar lines
             markeredgewidth=10,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,ar_h2o, marker = "o", markersize='25', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='cyan',
         linestyle = " ",label="Alt Radial",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, ar_h2o,  yerr = uncertainty_ar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,aar_h2o, marker = "o", markersize='20', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='green',linestyle = " ",label="Alt Along Radial",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, aar_h2o,  yerr = uncertainty_aar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,inner_h2o, marker = "o", markersize='15', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='red',linestyle = " ",label="Inner Positive",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, inner_h2o,  yerr = uncertainty_inner, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,outer_h2o, marker = "o", markersize='10', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='silver',linestyle = " ",label="Outer Positive",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, outer_h2o,  yerr = uncertainty_outer, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)


# Display the uncertainty percentage just below the points
'''
for i in range(len(p)):
    plt.text(p[i], aa_h2o[i] + aa_error[i] + 1.5, f'{uncertainty_percentage[i]:.2f}', 
             fontsize=24, color='k', ha='left',zorder=4)  # Adjust positioning with -1.5
'''
# Display the uncertainty percentage with a text box
# Set x-ticks at alternate positions
# Set alternate x-ticks by slicing every other tick
xticks_positions = (p[0], p[2], p[4])  # Selecting every other x position

#xticks_labels = [f'{int(val):.0f}' for val in p[::2]]  # Create labels for alternate ticks
#print(xticks_positions,xticks_labels)
# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)

plt.xlabel(r"Pressure (Pa) $\longrightarrow$", fontsize = "36")
#plt.title(r"Convergence test for 8.26 $\AA$ (Alt along) at 300 K and 1 bar",fontsize = "36")
plt.grid(visible=True,which='minor',color='#999999',linestyle='-',alpha=0.2)
#plt.ylabel("$CO_2$ adsorption in molecules/cylinder vol.($\AA^3$)", fontsize = "20")
plt.ylabel(r"$y_{H_{2}O}$ with $\sigma_y$ (mol/cell)", fontsize = "36")
plt.legend(loc="center left",prop={'size': 28})
plt.grid(True,axis='both',which='minor',linestyle='--', alpha=0.5,linewidth=3.5,zorder=1)
plt.grid(True,axis='both',which='major',linestyle='--', alpha=0.25,linewidth=2.75,zorder=1)
# Annotate with a box
plt.annotate(r'$|q| = 0.3915$,'+'\n'+'$d_{ICC} = 10.78 \, \AA$', xy=(0.05, 0.85), xycoords='axes fraction', 
             fontsize=28, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

plt.xscale('log')
plt.ylim(None,55.0)
#plt.yscale('log')
#plt.xlim(0.5,10)
#plt.ylim(0,36.0)
#plt.savefig('H2S_conv.png',dpi=500)
plt.show()
```

```python
z_length = 30.80e-8 #cm
r_eff_pore = (7.28/2)*1e-8 #cm
vpore = np.pi * z_length * (r_eff_pore)**2
#vpore = 1.65e-21 #cm3 
m_h2o = 2.99e-23 #g
rho_liquid_100kPa = 0.986525  #g/cm3
N_points = 6
#calculating density
rho_aa = (aa_h2o[-N_points:] * m_h2o)/vpore
rho_ar = (ar_h2o[-N_points:] * m_h2o)/vpore
rho_aar = (aar_h2o[-N_points:] * m_h2o)/vpore
rho_inner = (inner_h2o[-N_points:] * m_h2o)/vpore
rho_outer = (outer_h2o[-N_points:] * m_h2o)/vpore

#calculating the uncertainty
u_rho_aa = (uncertainty_aa[-N_points:] * m_h2o)/vpore
u_rho_ar = (uncertainty_ar[-N_points:] * m_h2o)/vpore
u_rho_aar = (uncertainty_aar[-N_points:] * m_h2o)/vpore
u_rho_inner = (uncertainty_inner[-N_points:] * m_h2o)/vpore
u_rho_outer = (uncertainty_outer[-N_points:] * m_h2o)/vpore

#CH4_Error = df.iloc[:,4].values
#print(Iast_co2)
#Doing the plotting
# set the font globally
#plt.rcParams.update({"font.family": "Times"})
marker_size = 55
fig1=plt.figure(dpi=800)
fig1.set_figwidth(15)
fig1.set_figheight(10)
xticks_positions = (p[0], p[2], p[3], p[4], p[5])  # Selecting every other x position
plt.rc('xtick',labelsize=48)
plt.rc('ytick',labelsize=48)
#plt.rcParams.update({'errorbar.capsize':3})
#plotting for 100 K
#plt.plot(Pressure100,GP_100, marker = "H",color="r",Linestyle="solid",label="GP (T = 100 K)")
# Plot the liquid density as a thick line across the pressure points
plt.plot(p, [rho_liquid_100kPa]*len(p), linewidth=5, \
         linestyle= '--',color='k', label=r'Bulk Water (100 kPa)')

plt.plot(p[-N_points:],rho_aa, marker = "o", markersize='27.5', 
         markeredgecolor='k',linewidth='10',color = "k", 
         markerfacecolor='orange',markeredgewidth='5',
         linestyle = " ",label="AA",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_aa,  yerr = u_rho_aa, 
             label="$\sigma_{(CFC-GCMC)}$",color = 'k', capsize=25,  # Length of the error bar caps
             elinewidth=10,  # Thickness of the error bar lines
             markeredgewidth=10,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_ar, marker = "o", markersize='25', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='cyan',
         linestyle = " ",label="AR",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_ar,  yerr = u_rho_ar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_aar, marker = "o", markersize='20', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='green',linestyle = " ",label="AAR",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_aar,  yerr = u_rho_aar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_inner, marker = "o", markersize='15', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='red',linestyle = " ",label="Inner",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_inner,  yerr = u_rho_inner, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_outer, marker = "o", markersize='10', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='silver',linestyle = " ",label="Outer",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_outer,  yerr = u_rho_outer, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)


# Display the uncertainty percentage just below the points
'''
for i in range(len(p)):
    plt.text(p[i], aa_h2o[i] + aa_error[i] + 1.5, f'{uncertainty_percentage[i]:.2f}', 
             fontsize=24, color='k', ha='left',zorder=4)  # Adjust positioning with -1.5
'''
# Display the uncertainty percentage with a text box
# Set x-ticks at alternate positions
# Set alternate x-ticks by slicing every other tick
# Ensure 'p' is a numpy array
# Define x-tick positions at logarithmic intervals


#ax.set_xticklabels([str(int(tick)) for tick in alternate_major_ticks])
# Set x-ticks at alternate positions
# Select every other pressure value for x-ticks
#xticks_positions = p[::1]  # Every other value in 'p'
#xticks_labels = [str(int(val)) for val in xticks_positions]  # Convert to strings for labels

# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)


#xticks_labels = [f'{int(val):.0f}' for val in p[::2]]  # Create labels for alternate ticks
#print(xticks_positions,xticks_labels)
# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)

plt.xlabel(r"Pressure (Pa) $\longrightarrow$", fontsize = "48")
#plt.title(r"Convergence test for 8.26 $\AA$ (Alt along) at 300 K and 1 bar",fontsize = "36")
plt.grid(visible=True,which='minor',color='#999999',linestyle='-',alpha=0.2)
#plt.ylabel("$CO_2$ adsorption in molecules/cylinder vol.($\AA^3$)", fontsize = "20")
plt.ylabel(r"${\rho}_{H_{2}O}$ (g/$cm^3$)", fontsize = "48")
plt.legend(loc="lower left",prop={'size': 32})
plt.grid(True,axis='both',which='minor',linestyle='--', alpha=0.5,linewidth=3.5,zorder=1)
plt.grid(True,axis='both',which='major',linestyle='--', alpha=0.25,linewidth=2.75,zorder=1)
# Annotate with a box
plt.annotate(r'$|q| = 0.7830$,'+'\n'+'$d_{ICC} = 10.78 \, \AA$', xy=(0.59, 0.25), xycoords='axes fraction', 
             fontsize=48, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

plt.xscale('log')
xticks_positions = [5.0*1e2, 1.0e3, 2.0e3, 4.0e3]  # Adjust these values as needed
# Define a function to format tick labels in scientific notation
def format_scientific_notation(val):
    exponent = int(np.floor(np.log10(val)))
    coeff = val / (10**exponent)
    if coeff == 1:
        return r'$10^{%d}$' % exponent
    else:
        return r'$%g \times 10^{%d}$' % (coeff, exponent)

# Create the labels using the formatting function
xticks_labels = [format_scientific_notation(val) for val in xticks_positions]
# Set the x-ticks and labels
# Set the x-ticks and labels
plt.xticks(xticks_positions, xticks_labels)
plt.ylim(0.25,1.25)
#plt.yscale('log')
plt.xlim(500,5000)
#plt.ylim(0,36.0)
#plt.savefig('H2S_conv.png',dpi=500)
plt.show()
```
