```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

#Reading the file
df = pd.read_csv('AA.csv',delimiter=',')
df2 = pd.read_csv('AR.csv',delimiter=',')
df3 = pd.read_csv('AAR.csv',delimiter=',')
df4 = pd.read_csv('Inner.csv',delimiter=',')
df5 = pd.read_csv('Outer.csv',delimiter=',')

#Reading the data
p = (df.iloc[:,0].values)
aa_n2o = df.iloc[:,1].values
aa_error = df.iloc[:,2].values
#2sigma
aa_error = 2*aa_error
uncertainty_aa = (aa_error) * 0.5

ar_n2o = df2.iloc[:,1].values
uncertainty_ar = df2.iloc[:,2].values

aar_n2o = df3.iloc[:,1].values
uncertainty_aar = df3.iloc[:,2].values

inner_n2o = df4.iloc[:,1].values
uncertainty_inner = df4.iloc[:,2].values

outer_n2o = df5.iloc[:,1].values
uncertainty_outer = df5.iloc[:,2].values

#CH4_Error = df.iloc[:,4].values
#print(Iast_co2)
#Doing the plotting
# set the font globally
#plt.rcParams.update({"font.family": "Times"})
marker_size = 55
fig1=plt.figure(dpi=800)
fig1.set_figwidth(15)
fig1.set_figheight(10)
plt.rc('xtick',labelsize=36)
plt.rc('ytick',labelsize=36)
#plt.rcParams.update({'errorbar.capsize':3})
#plotting for 100 K
#plt.plot(Pressure100,GP_100, marker = "H",color="r",Linestyle="solid",label="GP (T = 100 K)")

plt.plot(p,aa_n2o, marker = "o", markersize='27.5', 
         markeredgecolor='k',linewidth='10',color = "k", 
         markerfacecolor='orange',markeredgewidth='5',
         linestyle = " ",label="Alt Along",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, aa_n2o,  yerr = uncertainty_aa, 
             label="$\sigma_{CFC-GCMC}$",color = 'k', capsize=25,  # Length of the error bar caps
             elinewidth=10,  # Thickness of the error bar lines
             markeredgewidth=10,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,ar_n2o, marker = "o", markersize='25', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='cyan',
         linestyle = " ",label="Alt Radial",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, ar_n2o,  yerr = uncertainty_ar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,aar_n2o, marker = "o", markersize='20', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='green',linestyle = " ",label="Alt Along Radial",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, aar_n2o,  yerr = uncertainty_aar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,inner_n2o, marker = "o", markersize='15', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='red',linestyle = " ",label="Inner Positive",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, inner_n2o,  yerr = uncertainty_inner, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)

plt.plot(p,outer_n2o, marker = "o", markersize='10', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='silver',linestyle = " ",label="Outer Positive",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p, outer_n2o,  yerr = uncertainty_outer, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=5)


# Display the uncertainty percentage just below the points
'''
for i in range(len(p)):
    plt.text(p[i], aa_n2o[i] + aa_error[i] + 1.5, f'{uncertainty_percentage[i]:.2f}', 
             fontsize=24, color='k', ha='left',zorder=4)  # Adjust positioning with -1.5
'''
# Display the uncertainty percentage with a text box
# Set x-ticks at alternate positions
# Set alternate x-ticks by slicing every other tick
xticks_positions = (p[0], p[2], p[4])  # Selecting every other x position

#xticks_labels = [f'{int(val):.0f}' for val in p[::2]]  # Create labels for alternate ticks
#print(xticks_positions,xticks_labels)
# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)

plt.xlabel(r"Pressure (Pa) $\longrightarrow$", fontsize = "36")
#plt.title(r"Convergence test for 8.26 $\AA$ (Alt along) at 300 K and 1 bar",fontsize = "36")
plt.grid(visible=True,which='minor',color='#999999',linestyle='-',alpha=0.2)
#plt.ylabel("$CO_2$ adsorption in molecules/cylinder vol.($\AA^3$)", fontsize = "20")
plt.ylabel(r"$y_{H_{2}O}$ with $\sigma_y$ (mol/cell)", fontsize = "36")
plt.legend(loc="center left",prop={'size': 28})
plt.grid(True,axis='both',which='minor',linestyle='--', alpha=0.5,linewidth=3.5,zorder=1)
plt.grid(True,axis='both',which='major',linestyle='--', alpha=0.25,linewidth=2.75,zorder=1)
# Annotate with a box
plt.annotate(r'$|q| = 0.7830, d_{ICC} = 10.74 \, \AA$', xy=(0.05, 0.85), xycoords='axes fraction', 
             fontsize=28, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

plt.xscale('log')
plt.ylim(None,80.0)
#plt.yscale('log')
#plt.xlim(0.5,10)
#plt.ylim(0,36.0)
#plt.savefig('H2S_conv.png',dpi=500)
plt.show()
```

```python
z_length = 30.80e-8 #cm
r_eff_pore = (7.28/2)*1e-8 #cm
vpore = np.pi * z_length * (r_eff_pore)**2
#vpore = 1.65e-21 #cm3 
m_h2o = 2.99e-23 #g
rho_liquid_10kPa = 0.98660358  #g/cm3
N_points = 6
#calculating density
rho_aa = (aa_n2o[-N_points:] * m_h2o)/vpore
rho_ar = (ar_n2o[-N_points:] * m_h2o)/vpore
rho_aar = (aar_n2o[-N_points:] * m_h2o)/vpore
rho_inner = (inner_n2o[-N_points:] * m_h2o)/vpore
rho_outer = (outer_n2o[-N_points:] * m_h2o)/vpore

#calculating the uncertainty
u_rho_aa = (uncertainty_aa[-N_points:] * m_h2o)/vpore
u_rho_ar = (uncertainty_ar[-N_points:] * m_h2o)/vpore
u_rho_aar = (uncertainty_aar[-N_points:] * m_h2o)/vpore
u_rho_inner = (uncertainty_inner[-N_points:] * m_h2o)/vpore
u_rho_outer = (uncertainty_outer[-N_points:] * m_h2o)/vpore

#CH4_Error = df.iloc[:,4].values
#print(Iast_co2)
#Doing the plotting
# set the font globally
#plt.rcParams.update({"font.family": "Times"})
marker_size = 55
fig1=plt.figure(dpi=800)
fig1.set_figwidth(15)
fig1.set_figheight(10)
xticks_positions = (p[0], p[2], p[3], p[4], p[5])  # Selecting every other x position
plt.rc('xtick',labelsize=48)
plt.rc('ytick',labelsize=48)
#plt.rcParams.update({'errorbar.capsize':3})
#plotting for 100 K
#plt.plot(Pressure100,GP_100, marker = "H",color="r",Linestyle="solid",label="GP (T = 100 K)")
# Plot the liquid density as a thick line across the pressure points
plt.plot(p, [rho_liquid_10kPa]*len(p), linewidth=5, \
         linestyle= '--',color='k', label=r'Bulk Water (10 kPa)')

plt.plot(p[-N_points:],rho_aa, marker = "o", markersize='27.5', 
         markeredgecolor='k',linewidth='10',color = "k", 
         markerfacecolor='orange',markeredgewidth='5',
         linestyle = " ",label="AA",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_aa,  yerr = u_rho_aa, 
             label="$\sigma_{(CFC-GCMC)}$",color = 'k', capsize=25,  # Length of the error bar caps
             elinewidth=10,  # Thickness of the error bar lines
             markeredgewidth=10,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_ar, marker = "o", markersize='25', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='cyan',
         linestyle = " ",label="AR",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_ar,  yerr = u_rho_ar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_aar, marker = "o", markersize='20', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='green',linestyle = " ",label="AAR",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_aar,  yerr = u_rho_aar, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_inner, marker = "o", markersize='15', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='red',linestyle = " ",label="Inner",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_inner,  yerr = u_rho_inner, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)

plt.plot(p[-N_points:],rho_outer, marker = "o", markersize='10', 
         linewidth='10',color = "k", markeredgecolor='None', 
         markerfacecolor='silver',linestyle = " ",label="Outer",zorder=3)
#plt.fill_between(p, aa_h2s - (aa_error), aa_h2s + (aa_error),color='cyan', alpha=1.00,label="95% Confidence",zorder=1)

plt.errorbar(p[-N_points:], rho_outer,  yerr = u_rho_outer, 
             color = 'k', capsize=15,  # Length of the error bar caps
             elinewidth=5,  # Thickness of the error bar lines
             markeredgewidth=5,  # Thickness of the marker edges
             linewidth=0)


# Display the uncertainty percentage just below the points
'''
for i in range(len(p)):
    plt.text(p[i], aa_n2o[i] + aa_error[i] + 1.5, f'{uncertainty_percentage[i]:.2f}', 
             fontsize=24, color='k', ha='left',zorder=4)  # Adjust positioning with -1.5
'''
# Display the uncertainty percentage with a text box
# Set x-ticks at alternate positions
# Set alternate x-ticks by slicing every other tick
# Ensure 'p' is a numpy array
# Define x-tick positions at logarithmic intervals


#ax.set_xticklabels([str(int(tick)) for tick in alternate_major_ticks])
# Set x-ticks at alternate positions
# Select every other pressure value for x-ticks
#xticks_positions = p[::1]  # Every other value in 'p'
#xticks_labels = [str(int(val)) for val in xticks_positions]  # Convert to strings for labels

# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)


#xticks_labels = [f'{int(val):.0f}' for val in p[::2]]  # Create labels for alternate ticks
#print(xticks_positions,xticks_labels)
# Set the x-ticks to be displayed at alternate positions
#plt.xticks(xticks_positions, xticks_labels)

plt.xlabel(r"Pressure (Pa) $\longrightarrow$", fontsize = "48")
#plt.title(r"Convergence test for 8.26 $\AA$ (Alt along) at 300 K and 1 bar",fontsize = "36")
plt.grid(visible=True,which='minor',color='#999999',linestyle='-',alpha=0.2)
#plt.ylabel("$CO_2$ adsorption in molecules/cylinder vol.($\AA^3$)", fontsize = "20")
plt.ylabel(r"${\rho}_{H_{2}O}$ (g/$cm^3$)", fontsize = "48")
plt.legend(loc="lower left",prop={'size': 32})
plt.grid(True,axis='both',which='minor',linestyle='--', alpha=0.5,linewidth=3.5,zorder=1)
plt.grid(True,axis='both',which='major',linestyle='--', alpha=0.25,linewidth=2.75,zorder=1)
# Annotate with a box
plt.annotate(r'$|q| = 0.7830$,'+'\n'+'$d_{ICC} = 10.78 \, \AA$', xy=(0.575, 0.25), xycoords='axes fraction', 
             fontsize=48, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

plt.xscale('log')
xticks_positions = [5.0*1e2, 1.0e3, 2.0e3, 4.0e3]  # Adjust these values as needed
# Define a function to format tick labels in scientific notation
def format_scientific_notation(val):
    exponent = int(np.floor(np.log10(val)))
    coeff = val / (10**exponent)
    if coeff == 1:
        return r'$10^{%d}$' % exponent
    else:
        return r'$%g \times 10^{%d}$' % (coeff, exponent)

# Create the labels using the formatting function
xticks_labels = [format_scientific_notation(val) for val in xticks_positions]
# Set the x-ticks and labels
# Set the x-ticks and labels
plt.xticks(xticks_positions, xticks_labels)
plt.ylim(0.25,1.25)
#plt.yscale('log')
plt.xlim(500,5000)
#plt.ylim(0,36.0)
#plt.savefig('H2S_conv.png',dpi=500)
plt.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

# Reading the file
df = pd.read_csv('AA.csv', delimiter=',')
df2 = pd.read_csv('AR.csv', delimiter=',')
df3 = pd.read_csv('AAR.csv', delimiter=',')
df4 = pd.read_csv('Inner.csv', delimiter=',')
df5 = pd.read_csv('Outer.csv', delimiter=',')

# Reading the data
p = df.iloc[:, 0].values
aa_n2o = df.iloc[:, 1].values
aa_error = df.iloc[:, 2].values
# 2sigma
aa_error = 2 * aa_error
uncertainty_aa = (aa_error) * 0.5

ar_n2o = df2.iloc[:, 1].values
uncertainty_ar = df2.iloc[:, 2].values

aar_n2o = df3.iloc[:, 1].values
uncertainty_aar = df3.iloc[:, 2].values

inner_n2o = df4.iloc[:, 1].values
uncertainty_inner = df4.iloc[:, 2].values

outer_n2o = df5.iloc[:, 1].values
uncertainty_outer = df5.iloc[:, 2].values

# Create the figure
fig = go.Figure()

# Add the Alt Along data with error bars
fig.add_trace(go.Scatter(
    x=p, y=aa_n2o,
    mode='markers',
    marker=dict(size=20, color='orange', line=dict(width=5, color='black')),
    name="$N_{2}O$ (Alt Along)",
    error_y=dict(
        type='data',
        array=uncertainty_aa,
        visible=True,
        thickness=5,
        width=10,
        color='black'
    )
))

# Add the Alt Radial data with error bars
fig.add_trace(go.Scatter(
    x=p, y=ar_n2o,
    mode='markers',
    marker=dict(size=25, color='cyan'),
    name="$N_{2}O$ (Alt Radial)",
    error_y=dict(
        type='data',
        array=uncertainty_ar,
        visible=True,
        thickness=5,
        width=10,
        color='black'
    )
))

# Add the Alt Along Radial data with error bars
fig.add_trace(go.Scatter(
    x=p, y=aar_n2o,
    mode='markers',
    marker=dict(size=20, color='green'),
    name="$N_{2}O$ (Alt Along Radial)",
    error_y=dict(
        type='data',
        array=uncertainty_aar,
        visible=True,
        thickness=5,
        width=10,
        color='black'
    )
))

# Add the Inner Positive data with error bars
fig.add_trace(go.Scatter(
    x=p, y=inner_n2o,
    mode='markers',
    marker=dict(size=15, color='red'),
    name="$N_{2}O$ (Inner Positive)",
    error_y=dict(
        type='data',
        array=uncertainty_inner,
        visible=True,
        thickness=5,
        width=10,
        color='black'
    )
))

# Add the Outer Positive data with error bars
fig.add_trace(go.Scatter(
    x=p, y=outer_n2o,
    mode='markers',
    marker=dict(size=10, color='silver'),
    name="$N_{2}O$ (Outer Positive)",
    error_y=dict(
        type='data',
        array=uncertainty_outer,
        visible=True,
        thickness=5,
        width=10,
        color='black'
    )
))

# Update axis properties
fig.update_xaxes(type="log", title_text="Pressure (Pa)", title_font=dict(size=24))
fig.update_yaxes(title_text=r"$y_{N_{2}O}$ with $\sigma_y$ (mol/cell)", title_font=dict(size=24))

# Update layout with annotations
fig.update_layout(
    title="Convergence test for $N_{2}O$ adsorption",
    title_font_size=36,
    legend=dict(
        x=0.75,
        y=0.95,
        font=dict(size=20),
    ),
    xaxis=dict(tickfont=dict(size=18)),
    yaxis=dict(tickfont=dict(size=18)),
    width=1000,
    height=700,
    annotations=[
        dict(
            text=r"$|q| = 0.7830, d_{ICC} = 9.24 \, \AA$",
            xref="paper", yref="paper",
            x=0.05, y=0.75,
            showarrow=False,
            font=dict(size=20),
            bordercolor="black",
            borderwidth=2,
            borderpad=4,
            bgcolor="white"
        )
    ]
)

# Show the plot
fig.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reading the file
df = pd.read_csv('AA.csv', delimiter=',')
df2 = pd.read_csv('AR.csv', delimiter=',')
df3 = pd.read_csv('AAR.csv', delimiter=',')
df4 = pd.read_csv('Inner.csv', delimiter=',')
df5 = pd.read_csv('Outer.csv', delimiter=',')

# Reading the data
p = df.iloc[:, 0].values
aa_n2o = df.iloc[:, 1].values
aa_error = df.iloc[:, 2].values
# 2sigma
aa_error = 2 * aa_error
uncertainty_aa = (aa_error) * 0.5

ar_n2o = df2.iloc[:, 1].values
uncertainty_ar = df2.iloc[:, 2].values

aar_n2o = df3.iloc[:, 1].values
uncertainty_aar = df3.iloc[:, 2].values

inner_n2o = df4.iloc[:, 1].values
uncertainty_inner = df4.iloc[:, 2].values

outer_n2o = df5.iloc[:, 1].values
uncertainty_outer = df5.iloc[:, 2].values

# Set the aesthetic style of the plots
sns.set(style="whitegrid")

# Create the figure and axis
plt.figure(figsize=(15, 10), dpi=500)

# Plot using Seaborn lineplot with Matplotlib for error bars
sns.lineplot(x=p, y=aa_n2o, linewidth=3, color="orange", label="$N_{2}O$ (Alt Along)", zorder=3)
plt.errorbar(p, aa_n2o, yerr=uncertainty_aa, fmt='none', color='black', capsize=15, elinewidth=2)

sns.lineplot(x=p, y=ar_n2o, linewidth=3, color="cyan", label="$N_{2}O$ (Alt Radial)", zorder=3)
plt.errorbar(p, ar_n2o, yerr=uncertainty_ar, fmt='none', color='black', capsize=15, elinewidth=2)

sns.lineplot(x=p, y=aar_n2o, linewidth=3, color="green", label="$N_{2}O$ (Alt Along Radial)", zorder=3)
plt.errorbar(p, aar_n2o, yerr=uncertainty_aar, fmt='none', color='black', capsize=15, elinewidth=2)

sns.lineplot(x=p, y=inner_n2o, linewidth=3, color="red", label="$N_{2}O$ (Inner Positive)", zorder=3)
plt.errorbar(p, inner_n2o, yerr=uncertainty_inner, fmt='none', color='black', capsize=15, elinewidth=2)

sns.lineplot(x=p, y=outer_n2o, linewidth=3, color="silver", label="$N_{2}O$ (Outer Positive)", zorder=3)
plt.errorbar(p, outer_n2o, yerr=uncertainty_outer, fmt='none', color='black', capsize=15, elinewidth=2)

# Customizing the axis labels, grid, and legend
plt.xscale('log')
plt.xlabel(r"Pressure (Pa) $\longrightarrow$", fontsize=36)
plt.ylabel(r"$y_{N_{2}O}$ with $\sigma_y$ (mol/cell)", fontsize=36)
plt.xticks(fontsize=36)
plt.yticks(fontsize=36)

# Adding annotations
plt.annotate(r'$|q| = 0.7830, d_{ICC} = 9.24 \, \AA$', xy=(0.05, 0.75), xycoords='axes fraction', 
             fontsize=28, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.5'))

# Adding grid
plt.grid(True, axis='both', which='major', linestyle='--', alpha=0.25, linewidth=2.75, zorder=1)

# Adding the legend
plt.legend(loc="lower right", prop={'size': 28})

# Show plot
plt.show()
```
